```python
from langchain_groq import ChatGroq
model = ChatGroq(
    model="Gemma2-9b-It", groq_api_key=groq_api_key

)
model
```

```python
from langchain_core.messages import HumanMessage
model.invoke([HumanMessage(content = "Hi, My name is Dhanush and I am a AI Engineer")])
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}

def get_session_history(session_id: str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

with_message_history = RunnableWithMessageHistory(model, get_session_history)   
```

```python
config = {"configuration": {"session_id": "chat1"}}
```

```python
# Update config to use the correct key
config = {"configurable": {"session_id": "chat1"}}

response = with_message_history.invoke(
    [HumanMessage(content="Hi, My name is Dhanush and I am an AI Engineer")],
    config=config
)
```

```python
response.content
```

```python
with_message_history.invoke(
    [HumanMessage(content="Whats My name?")],
    config=config,
)
```

```python
from langchain_core.prompts import ChatPromptTemplate , MessagesPlaceholder
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a Helful Assistant. Answer all the Questions in {language}"),
        MessagesPlaceholder(variable_name="messages")
    ]
)
chain = prompt|model 
```

```python
response = chain.invoke({
    "messages": [HumanMessage(content = "Hi , My name is Dhanush")], "language": "Kannada"
})
response.content
```

```python
response.content
```

```python
with_message_history = RunnableWithMessageHistory(
    chain, 
    get_session_history,
    input_messages_key="messages"
)
```

```python
config = {"configurable": {"session_id": "chat3"}}
response = with_message_history.invoke(
    {
        "messages": [HumanMessage(content="Hi I am Dhanush")],
        "language": "Kannada"
    },
    config=config
)

response.content
```

```python
from langchain_core.messages import SystemMessage
from langchain_core.messages.utils import trim_messages

trimmer = trim_messages(
    max_tokens=70,
    strategy='last',
    token_counter=model,
    include_system=True,
    allow_partial=False,
    start_on="human"
)
messages = [
    SystemMessage(content="you are a good assistant"),
    HumanMessage(content="Hi, I am Bob"),
    AIMessage(content= "Hi!"),
    HumanMessage(content="I like a vanilla ice cream!"),
    AIMessage(content= "nice"),
    HumanMessage(content="2+2"),
    AIMessage(content= "4"),
    HumanMessage(content="Thanks!"),
    AIMessage(content= "no problem"),
]
trimmer.invoke(messages)
```
