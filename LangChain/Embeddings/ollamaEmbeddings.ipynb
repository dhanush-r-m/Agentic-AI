```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings = (
    OllamaEmbeddings(model = "llama2")
)
```

```python
embeddings
```

```python
r1 = embeddings.embed_documents(
    [
        "Alpha is the first letter of the Greek Alphabet.",
        "Beta is the second letter of the Greek Alphabet.",
    ]
) ## Here we can see our embbeding of ollama which are converted into vectors
```

```python
len(r1[0]) ## this is the length of the vector which is 4096
```

```python
embedings.embed_query("What is the first Letter of the greek alphabet?")
## VECTOR Representation of the embedding the model appears here from here we can have the markdown zone.
```
