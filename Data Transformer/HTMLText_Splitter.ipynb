```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>Sample HTML Page</title>
    <style>
        body {
            font-family: Arial, sans-serif;
            margin: 40px;
            background-color: #f2f2f2;
        }
        h1 {
            color: #333366;
        }
        p {
            color: #444;
        }
    </style>
</head>
<body>
    <h1>Welcome to My Sample Page</h1>
    <p>This is a simple HTML page with a heading, paragraph, image, and a link.</p>
    
    <img src="https://via.placeholder.com/300x200" alt="Sample Image" width="300">
    
    <p>Visit <a href="https://www.example.com" target="_blank">Example.com</a> for more information.</p>
</body>
</html>
"""

headers_to_split_on = [
    ("h1", "Header_1"),
    ("h2", "Header_2"),
    ("h3", "Header_3")
]

html_splitter = HTMLHeaderTextSplitter(headers_to_split_on)
html_headers_splits = html_splitter.split_text(html_string)
html_headers_splits
```
