```python
%pip install langchain-community

from langchain_community.document_loaders import TextLoader
from langchain.text_splitter import CharacterTextSplitter
from langchain_community.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
loader = TextLoader("speech.txt")
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap = 30)
docs = text_splitter.split_documents(documents)
```

```python
docs
```

```python
embeddings = OllamaEmbeddings()
db = FAISS.from_documents(docs, embeddings)
db
```

```python
query = "What is the main topic of this sppech?"
docs = db.similarity_search(query)
docs
```

```python
retriever = db.as_retriever() ## retriever is used as the operation to retrieve documents based on a query
retriever.invoke(query)
docs[0].page_content
```

```python
docs_and_sscores = db.similarity_search_with_score(query)
docs_and_sscores ## this returns the documents along with their similarity scores and scoring is done by distance query between the query and the document embeddings using the manhattan distance metric
```

```python
embedding_vector = embeddings.embed_query(query)
embediing_vector
```

```python
docs_score = db.similarity_search_by_vector(embedding_vector)
docs_score
```

```python
db.save_local("faiss_index")
```
