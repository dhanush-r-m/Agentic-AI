```python
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.emmbeddings import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharcterTextSplitter
```

```python
loader = Textloader("speech.txt")
data = loader.load()
data
```

```python
text_splitter = RecursiveCharcterTextSplitter(chunk_size=1000, chunk_overlap=0)
splits = text_splitter.split_documents(data)
```

```python
embeddings = OllamaEmbeddings()
vectordb = Chroma.from_documents(documents=splits, embedding = embeddings)
vectordb
```

```python
query = "what is the sppech about"
docs = vectordb.similarity_search(query)
docs[0].page_content
```

```python
vectordb = Chroma.from_documents(documents=splits, embeddings = embeddings, persist_directory = "./chroma_db")
```

```python
db2 = Chroma(persist_directory = "./chroma_db" , embedding_function = embedings)
docs = db2.similarity_search(query)
print(docs[0].page_content)
```

```python
docs = vectordb.similarity_search_with_score(query)
docs
```

```python
retriever = vectordb.as_reteriever()
retriever.invoke(query)[0].page_content
```
