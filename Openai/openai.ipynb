```python
import os
from dotenv import load_dotenv
load_dotenv() ## use your .env file to set environment variables
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TACing_V2"] = true

os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4o", temperature=0.7, max_tokens=1000, streaming=True, verbose=True)
print(llm)
```

```python
## Input and get respone from the LLM
result = llm.invoke("What is Generative AI?")
```

```python
print(result)
```

```python
from langchain_core.prompt import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_messages([
    (
        "system", "You are a AI Expert Engineer, Provide me answer based on the question."
        "user", "{input}"
    )
    ])
prompt
```

```python
chain = prompt | llm

response = chain.invoke({"input": "What is the about Langchain?"})
print(response)
```

```python
type(response)
```

```python
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt | llm | output_parser
response = chain.invoke({"input": "What is the about Langchain?"})
print(response)
```
