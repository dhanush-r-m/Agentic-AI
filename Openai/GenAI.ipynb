```python
## Data - Ingestion -- From the website we need to scrape the data
from langchain_community.document_loaders import WebBaseLoader
```

```python
loader = WebBaseLoader("https://python.langchain.com/docs/get_started/introduction.html")
loader
```

```python
docs = loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterSplitter(chunk_size = 1000, chunk_overlap = 200)
documents = text_splitter.split_documents(docs)
```

```python
from langchain_openai impor OpenAIEmbeddings
embeddings = OpenAIembeddings()
```

```python
from lanchain_community.vectorstores import FAISS
vectorstore = FAISS.from_documents(documents, embeddings)
```
