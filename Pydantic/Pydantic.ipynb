```python
from typing import Optional

class Person(BaseModel):
    name :str
    age : int
    city : str
    salary : Optional[float] = None
    is_active : bool = True

person = Person(name ="Dhanush" , age = 19 , city ="Bangalore", salary = 50000.0, is_active = True)
print(person)
```

```python
from pydantic import BaseModel
from typing import List, Optional

class classroom(BaseModel):
    room_no : int
    name : List[str]
    capacity : int
```

```python
#create classroom
classroom1 = classroom(room_no =101 , name =["Dhanush" , "John" , "doe"] , capacity = 30)
print(classroom1)
```

```python
from pydantic import BaseModel, Field
class Item(BaseModel):
    name : str = Field(min_length=3, max_length=50)
    price :float = Field(gt = 0, le = 1000) #greater than 0 and less than or equal to 1000
    quantity : int = Field(ge = 1)

item = Item(name = "Apple" , price = 10.5, quantity = 25)
print(item)
```

```python
class User(BaseModel):
    username: str = Field(..., description="The username of the user")
    email: str = Field(default_factory=lambda: "user@example.com", description="The email of the user")
    age: int = Field(gt=0, description="The age of the user, must be greater than 0")

user = User(username="john_doe", age=30)
print(user)
```

```python
print(user.schema())
```

```python
print(user.model_json_schema())
```
